3. **Data Splits**: 70% Train, 15% Validation, 15% untouched Test set (`data/processed/train.csv`, `val.csv`, `test.csv`).
4. **Artifacts Exported**:
   - `models/classifier.joblib`
   - `models/label_mapping.json`
   - `reports/baseline_training_summary.json`
